Struktur prompt dapat disusun melalui komponen Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Verification (R–T–C–I–K–O–Q–V) yang mementingkan kejelasan, relevansi, konsistensi, dan sejauh mana instruksi tersebut mengurangi keputusan penting yang harus ditebak oleh AI. ### **KASUS 1 — ANALISIS ANOMALI MEDAN MAGNET (SURVEI MAGNETIK)**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang belajar mengolah dan menginterpretasikan data anomali medan magnet hasil survei magnetik darat.

#### **Task**

Analisis peta anomali medan magnet total (atau anomali setelah reduksi ke kutub) untuk mengidentifikasi zona anomali positif dan negatif, memperkirakan pola sebaran benda termagnetisasi, serta mengidentifikasi kemungkinan struktur geologi yang terlihat dari pola anomali.

#### **Context**

Metode magnetik memanfaatkan variasi suseptibilitas magnetik material bawah permukaan untuk memperoleh informasi mengenai keberadaan benda atau struktur geologi tertentu, misalnya intrusi batuan beku, benda terkubur, atau zona sesar. Nilai anomali magnetik dipengaruhi oleh kontras suseptibilitas, kedalaman sumber, arah kemagnetan bumi, dan bentuk benda anomali sehingga pola yang dipetakan dapat berupa dipol (kutub positif dan negatif) sebelum dilakukan reduksi ke kutub.

Interpretasi pada tahap ini bersifat kualitatif berdasarkan pola anomali yang terlihat pada peta, sehingga belum tentu setiap anomali dapat langsung dikaitkan dengan jenis batuan atau benda tertentu tanpa data pendukung lain.

#### **Input**

Data yang digunakan berupa:

* Peta atau grid nilai anomali medan magnet total (nT);
* Koordinat titik pengukuran (x, y) atau lintang/bujur;
* Informasi apakah data sudah melalui koreksi IGRF dan reduksi ke kutub;
* Informasi tambahan mengenai lokasi survei atau geologi regional jika tersedia.

#### **Constraints**

1. Jangan menentukan jenis benda atau batuan penyebab anomali hanya berdasarkan besar nilai anomali.
2. Gunakan pola anomali yang benar-benar terlihat pada peta sebagai dasar interpretasi.
3. Bedakan antara pengamatan langsung terhadap pola anomali dan interpretasi geologi.
4. Jangan menyatakan kedalaman atau geometri sumber anomali secara pasti tanpa dilakukan pemodelan lebih lanjut.
5. Jangan mengubah atau menghilangkan bagian data anomali.
6. Jika pola anomali masih berbentuk dipol dan belum direduksi ke kutub, nyatakan keterbatasan tersebut sebelum melakukan interpretasi posisi sumber.
7. Jangan mengabaikan kemungkinan noise akibat gangguan magnetik buatan (kultur budaya) di sekitar area survei.

#### **Output**

Hasil yang direncanakan berupa:

1. identifikasi zona anomali magnetik tinggi dan rendah
2. deskripsi pola sebaran anomali (memanjang, membulat, atau tidak beraturan)
3. perkiraan arah atau tren struktur berdasarkan pola anomali
4. identifikasi kemungkinan sumber anomali, jika terdapat indikasi yang cukup
5. alasan yang mendukung setiap interpretasi
6. tabel yang memisahkan pengamatan pola anomali dan interpretasi geologi
7. kesimpulan singkat mengenai pola magnetik bawah permukaan.

#### **Quality Criteria**

Hasil harus:

* Berdasarkan pola anomali yang benar-benar terlihat pada peta
* Mempertimbangkan bentuk dan orientasi anomali
* Mempertimbangkan status koreksi data (IGRF, reduksi ke kutub)
* Mencantumkan satuan nilai anomali (nT)
* Membedakan fakta pengamatan dari dugaan interpretasi
* Menyatakan ketidakpastian apabila bukti tidak cukup

#### **Verification**

Periksa kembali seluruh peta anomali setelah interpretasi dilakukan. Pastikan setiap zona anomali yang disebutkan memiliki dasar visual pada peta. Periksa pula apakah pola anomali konsisten dengan area di sekitarnya dan apakah terdapat kemungkinan pengaruh gangguan magnetik buatan yang belum dipertimbangkan. Jika suatu interpretasi tidak dapat diverifikasi dari data, tandai sebagai tidak pasti dan jangan menyajikannya sebagai kesimpulan pasti.

#### **Rencana hasil**

Output yang diharapkan bukan berupa penentuan jenis benda penyebab anomali secara pasti, melainkan peta zona anomali, deskripsi pola sebaran, dan kemungkinan struktur berdasarkan bukti yang terlihat pada data.


### **KASUS 2 — ANALISIS ANOMALI GRAVITASI BOUGUER**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang belajar mengolah dan menginterpretasikan peta anomali gravitasi Bouguer.

#### **Task**

Analisis peta anomali Bouguer untuk mengidentifikasi zona anomali gravitasi tinggi dan rendah serta melihat kemungkinan batas atau kontras densitas bawah permukaan berdasarkan pola sebaran anomali.

#### **Context**

Metode gravitasi memanfaatkan variasi densitas material bawah permukaan untuk memperoleh informasi mengenai struktur geologi, misalnya cekungan sedimen, intrusi, atau struktur sesar. Anomali Bouguer diperoleh setelah data gravitasi terukur dikoreksi terhadap lintang, ketinggian (free air dan Bouguer), serta topografi di sekitar titik pengukuran. Nilai anomali yang tinggi umumnya berasosiasi dengan massa berdensitas lebih besar, sedangkan nilai rendah berasosiasi dengan massa berdensitas lebih kecil, tetapi hubungan ini tidak selalu tunggal karena efek superposisi dari berbagai sumber pada kedalaman berbeda.

#### **Input**

Input terdiri atas:

* Data anomali Bouguer (mGal) pada tiap titik pengukuran atau dalam bentuk grid;
* Koordinat titik pengukuran;
* Informasi densitas reduksi yang digunakan dalam koreksi Bouguer;
* Informasi topografi daerah penelitian jika tersedia;
* Jika tersedia, informasi geologi regional daerah penelitian.

#### **Constraints**

1. Jangan langsung menyatakan suatu nilai anomali sebagai jenis batuan atau struktur tertentu.
2. Jangan mengubah data anomali Bouguer asli.
3. Bedakan anomali hasil pengukuran/koreksi dari hasil interpretasi.
4. Jangan menyatakan keberadaan struktur geologi tertentu jika tidak terdapat pola anomali yang mendukung.
5. Jika dilakukan pemisahan anomali regional dan residual, jelaskan metode yang digunakan dan jangan menganggap hasil pemisahan sebagai representasi pasti kondisi bawah permukaan.
6. Perhatikan satuan anomali (mGal) secara konsisten.
7. Jangan mengabaikan kemungkinan pengaruh topografi yang belum terkoreksi dengan baik.

#### **Output**

Hasil yang direncanakan berupa:

1. identifikasi zona anomali gravitasi rendah
2. identifikasi zona anomali gravitasi tinggi
3. deskripsi pola sebaran anomali secara lateral
4. kemungkinan batas kontras densitas antar-zona
5. kemungkinan interpretasi struktur geologi berdasarkan pola anomali
6. tabel zona, posisi, nilai anomali, pengamatan, dan interpretasi
7. kesimpulan mengenai pola gravitasi bawah permukaan.

#### **Quality Criteria**

Analisis harus:

* Menggunakan nilai anomali dari data
* Mempertahankan satuan mGal
* Membedakan observasi dan interpretasi
* Memperhatikan posisi spasial titik pengukuran
* Tidak menghubungkan nilai anomali dengan jenis batuan secara berlebihan
* Menjelaskan dasar dari setiap interpretasi

#### **Verification**

Periksa kembali peta anomali Bouguer secara keseluruhan. Pastikan zona yang disebut sebagai anomali tinggi atau rendah memang sesuai dengan rentang nilai pada data. Periksa pula apakah interpretasi suatu zona konsisten dengan pola di sekitarnya dan apakah terdapat bagian yang memiliki kerapatan data pengukuran rendah sehingga interpolasi kurang dapat diandalkan.

#### **Rencana hasil**

Output yang diharapkan berupa peta yang telah diidentifikasi zona anomalinya, tabel karakteristik tiap zona, serta interpretasi awal mengenai kemungkinan kontras densitas bawah permukaan, bukan penentuan jenis batuan atau struktur secara pasti tanpa data pendukung tambahan.


### **KASUS 3 — ANALISIS SEBARAN HIPOSENTER GEMPA BUMI (SEISMISITAS)**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari analisis data seismisitas untuk melihat pola kegempaan suatu wilayah.

#### **Task**

Analisis data hiposenter gempa bumi (lintang, bujur, kedalaman, magnitudo, waktu kejadian) untuk mengidentifikasi pola sebaran spasial dan kedalaman gempa serta mengetahui apakah pola tersebut menunjukkan kecenderungan mengelompok pada zona atau struktur tertentu.

#### **Context**

Data seismisitas mencatat parameter hiposenter dari hasil relokasi gempa oleh jaringan seismograf. Sebaran hiposenter dapat memberikan gambaran mengenai keberadaan sesar aktif, zona subduksi, atau zona sesar lain, tetapi keakuratan pola sebaran sangat bergantung pada kualitas relokasi, cakupan jaringan stasiun, dan ketidakpastian penentuan kedalaman. Pola pengelompokan gempa (klaster) tidak selalu menunjukkan struktur tunggal dan perlu dibedakan dari sebaran acak atau efek jaringan pemantauan yang tidak merata.

#### **Input**

Input berupa:

* Data katalog gempa: waktu kejadian, lintang, bujur, kedalaman, magnitudo;
* Periode waktu pengamatan;
* Wilayah kajian;
* Informasi ketidakpastian lokasi/kedalaman jika tersedia;
* Jika tersedia, informasi sesar aktif atau zona subduksi pada wilayah kajian.

#### **Constraints**

1. Jangan menyatakan keberadaan sesar aktif tertentu hanya berdasarkan pengelompokan titik hiposenter tanpa data pendukung lain.
2. Jangan mengabaikan ketidaksetaraan sebaran akibat perbedaan kerapatan jaringan stasiun pemantau.
3. Pertahankan satuan magnitudo, kedalaman (km), dan koordinat asli.
4. Jangan mengubah atau menghapus data gempa dengan magnitudo kecil tanpa menjelaskan alasan penyaringan.
5. Bedakan pola sebaran hasil pengamatan dari interpretasi tektonik.
6. Jika data kedalaman memiliki ketidakpastian besar (misalnya kedalaman default), nyatakan keterbatasan tersebut.
7. Jangan menyimpulkan tren temporal (peningkatan/penurunan aktivitas) dari periode data yang terlalu singkat.

#### **Output**

Hasil yang direncanakan berupa:

1. peta sebaran episenter gempa
2. penampang kedalaman hiposenter (depth section)
3. identifikasi pola pengelompokan atau tren spasial
4. hubungan magnitudo dan kedalaman jika relevan
5. kemungkinan asosiasi pola sebaran dengan struktur tektonik, jika terdapat indikasi
6. tabel ringkasan statistik (jumlah kejadian, rentang magnitudo, rentang kedalaman)
7. kesimpulan singkat mengenai pola seismisitas wilayah kajian.

#### **Quality Criteria**

Hasil harus:

* Menggunakan seluruh rentang magnitudo dan kedalaman yang tersedia atau menjelaskan kriteria penyaringan
* Mempertahankan satuan dan sistem koordinat
* Membedakan pola pengamatan dari interpretasi tektonik
* Mempertimbangkan keterbatasan jaringan pemantauan
* Menyatakan ketidakpastian lokasi/kedalaman apabila relevan
* Menjelaskan dasar dari setiap interpretasi pola

#### **Verification**

Periksa kembali sebaran episenter dan hiposenter terhadap keseluruhan data. Pastikan klaster yang disebutkan benar-benar konsisten secara spasial dan tidak semata-mata disebabkan oleh kerapatan stasiun pemantau yang tidak merata. Periksa pula kewajaran rentang magnitudo dan kedalaman yang digunakan sebelum menarik kesimpulan.

#### **Rencana hasil**

Output yang diharapkan berupa gambaran pola sebaran seismisitas beserta kemungkinan asosiasinya dengan struktur tektonik, bukan penentuan sesar aktif secara pasti tanpa data geologi atau geodetik pendukung.


### **KASUS 4 — ANALISIS LOG SUMUR (WELL LOG) UNTUK IDENTIFIKASI LITOLOGI**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari analisis data log sumur untuk keperluan interpretasi litologi dan zona reservoir secara kualitatif.

#### **Task**

Analisis data log sumur (gamma ray, resistivitas, densitas, dan/atau neutron porosity) untuk mengidentifikasi kemungkinan litologi tiap interval kedalaman serta menandai zona yang berpotensi sebagai reservoir berdasarkan karakter log.

#### **Context**

Log sumur merekam sifat fisik batuan secara kontinu terhadap kedalaman, misalnya radioaktivitas alami (gamma ray), resistivitas formasi, dan densitas batuan. Kombinasi beberapa jenis log dapat digunakan untuk membedakan litologi secara kualitatif, misalnya batupasir, batulempung, atau batugamping, serta mengindikasikan kandungan fluida secara umum. Namun, interpretasi litologi dari log bersifat tidak unik karena nilai log yang serupa dapat dihasilkan oleh litologi yang berbeda, sehingga diperlukan kehati-hatian dan idealnya dikonfirmasi dengan data core atau cutting jika tersedia.

#### **Input**

Input berupa:

* Data log gamma ray (API) terhadap kedalaman;
* Data log resistivitas (ohm.m) terhadap kedalaman;
* Data log densitas dan/atau neutron porosity jika tersedia;
* Informasi kedalaman (MD/TVD) dan satuan yang digunakan;
* Informasi zona target atau formasi jika tersedia.

#### **Constraints**

1. Jangan menentukan jenis litologi secara pasti hanya dari satu jenis log tanpa mempertimbangkan log lainnya.
2. Jangan mengubah nilai log asli atau melakukan smoothing tanpa menjelaskan metode dan alasannya.
3. Bedakan pembacaan nilai log (observasi) dari interpretasi litologi.
4. Jangan menyatakan suatu interval sebagai reservoir produktif hanya berdasarkan log tanpa mempertimbangkan data uji produksi atau core.
5. Pertahankan satuan tiap jenis log (API, ohm.m, g/cc, dan lainnya).
6. Jika terdapat data yang hilang (gap) pada interval tertentu, jangan mengisi otomatis tanpa menjelaskan metode.
7. Jangan mengabaikan kemungkinan pengaruh lubang bor (washout) atau kondisi lingkungan pengukuran terhadap kualitas log.

#### **Output**

Hasil yang direncanakan berupa:

1. plot log (gamma ray, resistivitas, dan log pendukung lainnya) terhadap kedalaman
2. identifikasi interval dengan karakter litologi berbeda
3. perkiraan jenis litologi kualitatif tiap interval
4. identifikasi zona yang berpotensi sebagai reservoir berdasarkan kombinasi log
5. alasan yang mendukung setiap interpretasi interval
6. tabel interval kedalaman, karakter log, dan interpretasi litologi
7. kesimpulan singkat mengenai profil litologi sumur.

#### **Quality Criteria**

Hasil harus:

* Menggunakan kombinasi lebih dari satu jenis log jika tersedia
* Mempertahankan satuan dan kedalaman asli
* Membedakan pembacaan log dari interpretasi litologi
* Menyatakan ketidakpastian apabila karakter log ambigu
* Menjelaskan dasar penentuan batas antar-interval
* Tidak mengklaim zona sebagai reservoir produktif tanpa data pendukung tambahan

#### **Verification**

Periksa kembali seluruh interval kedalaman setelah interpretasi dilakukan. Pastikan batas litologi yang ditentukan konsisten dengan perubahan nilai log yang terlihat. Periksa pula apakah terdapat interval dengan kualitas data log yang buruk (misalnya akibat washout) yang perlu ditandai sebagai tidak pasti sebelum interpretasi litologi disajikan sebagai kesimpulan.

#### **Rencana hasil**

Output yang diharapkan berupa profil litologi kualitatif per interval kedalaman beserta indikasi zona berpotensi reservoir, bukan penentuan jenis batuan atau kandungan fluida secara pasti tanpa data core, cutting, atau uji produksi.


### **KASUS 5 — ANALISIS DEFORMASI PERMUKAAN TANAH DARI DATA GPS/InSAR**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari analisis data geodetik untuk memantau deformasi permukaan tanah menggunakan data GPS kontinu dan/atau InSAR.

#### **Task**

Analisis data perpindahan posisi (time series GPS) dan/atau peta deformasi hasil InSAR untuk mengidentifikasi pola dan laju deformasi permukaan tanah pada wilayah kajian, termasuk kemungkinan penyebab umum seperti aktivitas vulkanik, tektonik, atau penurunan tanah, apabila terdapat indikasi yang cukup.

#### **Context**

Data GPS kontinu mencatat perubahan posisi titik pengamatan terhadap waktu pada komponen horizontal (utara-timur) dan vertikal, sedangkan InSAR memberikan informasi perpindahan permukaan dalam arah line-of-sight satelit pada cakupan spasial yang lebih luas. Pola deformasi dapat berasosiasi dengan berbagai proses geofisika, misalnya inflasi/deflasi tubuh vulkanik, akumulasi regangan antar-gempa, atau penurunan tanah akibat ekstraksi air tanah maupun beban bangunan. Interpretasi penyebab deformasi memerlukan kehati-hatian karena pola yang tampak serupa dapat dihasilkan oleh proses yang berbeda.

#### **Input**

Input berupa:

* Data time series posisi GPS (komponen utara, timur, vertikal) terhadap waktu;
* Peta deformasi InSAR (line-of-sight displacement) jika tersedia;
* Koordinat dan periode pengamatan tiap titik/piksel;
* Informasi baseline atau referensi yang digunakan;
* Jika tersedia, informasi geologi, aktivitas vulkanik, atau kondisi hidrogeologi wilayah kajian.

#### **Constraints**

1. Jangan menentukan penyebab deformasi (vulkanik, tektonik, atau antropogenik) hanya berdasarkan pola perpindahan tanpa mempertimbangkan konteks geologi wilayah.
2. Jangan mengubah data posisi atau nilai displacement asli.
3. Bedakan noise/gangguan pengukuran (misalnya akibat atmosfer pada InSAR atau multipath pada GPS) dari sinyal deformasi yang sebenarnya.
4. Pertahankan satuan perpindahan (mm atau cm) dan periode waktu pengamatan.
5. Jangan mengisi data yang hilang pada time series secara otomatis tanpa menjelaskan metode.
6. Jika data InSAR hanya menunjukkan komponen line-of-sight, jangan langsung menyimpulkan arah pergerakan tiga dimensi tanpa data pendukung tambahan.
7. Jangan menyimpulkan percepatan atau perlambatan deformasi dari periode data yang terlalu singkat.

#### **Output**

Hasil yang direncanakan berupa:

1. grafik time series perpindahan GPS per komponen
2. peta pola deformasi dari InSAR jika tersedia
3. estimasi laju deformasi (mm/tahun) pada titik atau zona yang diamati
4. identifikasi area dengan laju deformasi yang menonjol
5. kemungkinan asosiasi pola deformasi dengan proses geofisika tertentu, jika terdapat indikasi
6. tabel titik/zona, laju deformasi, dan interpretasi
7. kesimpulan singkat mengenai pola deformasi wilayah kajian.

#### **Quality Criteria**

Hasil harus:

* Mempertahankan satuan perpindahan dan periode waktu
* Membedakan sinyal deformasi dari noise pengukuran
* Menjelaskan metode estimasi laju deformasi yang digunakan
* Tidak memberikan interpretasi penyebab deformasi yang tidak didukung data
* Menyatakan keterbatasan apabila hanya tersedia komponen line-of-sight
* Menjelaskan dasar dari setiap interpretasi

#### **Verification**

Periksa kembali time series dan/atau peta deformasi secara keseluruhan. Pastikan pola atau laju deformasi yang disebutkan konsisten dengan data pada periode dan lokasi yang berdekatan. Periksa pula kemungkinan pengaruh noise atmosfer, multipath, atau kesalahan referensi sebelum menyatakan suatu pola sebagai sinyal deformasi yang nyata.

#### **Rencana hasil**

Output yang diharapkan berupa gambaran pola dan laju deformasi permukaan tanah pada wilayah kajian, bukan penentuan penyebab deformasi secara pasti tanpa data geologi, hidrogeologi, atau geodetik pendukung tambahan.
